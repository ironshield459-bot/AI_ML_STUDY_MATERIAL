# Python Dictionaries – Common & Important Concepts Only

This notebook covers **only the most common and important concepts** of Python **dictionaries**.

Dictionaries are used everywhere in Python for:
- Fast lookups
- Structured data
- Backend APIs & JSON handling
- Real-world problem solving

## 1. What is a Dictionary?

A **dictionary (`dict`)** is a:
- Mutable data type
- Collection of **key–value** pairs
- Optimized for fast lookup by key (≈ O(1))

**Important rules:**
- Keys must be **hashable** (immutable)
- Values can be any data type
- Keys are unique
- Insertion order is preserved (Python 3.7+)

```python
d = {
    "name": "Danish",
    "age": 21,
    "is_student": True
}
print(d)
```

## 2. Creating Dictionaries (Common Ways)

```python
d1 = {}
d2 = dict()
d3 = dict(name="Alex", age=20)
d4 = dict([("a", 1), ("b", 2)])

print(d1)
print(d2)
print(d3)
print(d4)
```

## 3. Accessing & Modifying Values

Use `get()` for **safe access**.

```python
user = {"name": "Alex", "age": 21}

print(user["name"])
print(user.get("salary", 0))  # safe access

user["age"] = 22
user["city"] = "Delhi"
print(user)
```

## 4. Commonly Used Dictionary Methods

```python
d = {"a": 1, "b": 2}

print(d.keys())
print(d.values())
print(d.items())
```

```python
d.update({"b": 20, "c": 3})
print(d)

print(d.pop("a"))
print(d)
```

```python
d.setdefault("role", "user")
print(d)
```

## 5. Membership Testing

`in` checks **keys only**.

```python
print("age" in user)
print("Alex" in user)
```

## 6. Iterating Over Dictionaries

```python
for key, value in user.items():
    print(key, value)
```

## 7. Dictionary Comprehension (Important Skill)

```python
squares = {x: x*x for x in range(5)}
print(squares)
```

## 8. Nested Dictionaries (Very Common)

```python
student = {
    "name": "Alex",
    "marks": {"math": 90, "cs": 95}
}

print(student["marks"]["cs"])
```
