# Python Sets – Common & Important Concepts Only

This notebook covers **only the common, important, and practical concepts** of Python **sets**.

Sets are mainly used for:
- Removing duplicates
- Fast membership testing
- Mathematical set operations (union, intersection, etc.)

## 1. What is a Set?

A **set** is a:
- Unordered collection
- Mutable container
- Unindexed data type
- Collection of **unique** elements

**Important rules:**
- Duplicate values are automatically removed
- Elements must be **hashable** (immutable types)

```python
s = {1, 2, 3, 3}
print(s)
```

## 2. Creating Sets (Common Ways)

⚠️ `{}` creates a dictionary, not a set.

```python
s1 = set()
s2 = {1, 2, 3}
s3 = set([1, 2, 2, 3])
s4 = set("banana")

print(s1)
print(s2)
print(s3)
print(s4)
```

## 3. Mutability Rules (Very Important)

- Set itself is mutable
- Elements inside a set must be immutable

```python
s = {1, 2}
s.add(3)
print(s)

# s.add([4, 5])  # ❌ list is not hashable
```

## 4. Commonly Used Set Methods

Only the methods that are used most often in real code.

```python
s = {1, 2}

s.add(3)
print(s)

s.update([4, 5])
print(s)

s.remove(2)
print(s)

s.discard(10)  # no error if missing
print(s)
```

### pop() and clear()

```python
s = {1, 2, 3}
print(s.pop())  # arbitrary element
print(s)

s.clear()
print(s)
```

## 5. Set Operations (Core Concept)

These are the **main reason sets exist**.

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference
print(a ^ b)   # symmetric difference
```

## 6. Membership Testing (Very Fast)

```python
nums = {1, 2, 3}
print(2 in nums)
print(5 in nums)
```

## 7. Built-in Functions Used with Sets

```python
s = {10, 20, 30}
print(len(s))
print(max(s))
print(min(s))
print(sum(s))
print(sorted(s))
```

## 8. frozenset (Immutable Set)

Used when you need a set that cannot change.

```python
fs = frozenset([1, 2, 3])
print(fs)
```
