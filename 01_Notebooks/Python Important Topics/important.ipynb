# Packing and Unpacking 

Packing means collecting multiple values into a single variable or container (usually a tuple or dictionary).

Unpacking means extracting values from a container and spreading them into separate variables or arguments.

## 1. * Operator for Packing and Unpacking

### Packing with * Operator

```python
def sum(*arr):
    sum = 0
    for num in arr:
        sum += num
    return sum

result = sum(1,2,3,4,5)
print(result)
```

### Unpacking with * Operator

```python
def sum(a,b,c):
    result = a + b + c
    return result

nums = [1,2,3]

result = sum(*nums)
print(result)
```

## 2. ** Operator for Packing and Unpacking

### Packing with ** Operator

```python
def print_user_data(**user_data):
    for key,value in user_data.items():
        print(f"key:{key} , value:{value}")

print_user_data(user_id=1029484,name="Danish",age="23",city="New Delhi",is_logged_in = True)
```

### Unpacking with ** Operator

```python
def print_user_data(user_id,name,age,city,is_logged_in):
    print(f"{name} lives in {city}.")

user_data = {
    "name" : "Danish",
    "city" : "New Delhi",
    "age" : 23,
    "user_id" : 3819494092,
    "is_logged_in" : False
}

print_user_data(**user_data)
```

# *args and **kwargs 
### A standard and popular way of packing, already covered above

## 1. ***args** (Variable Positional Arguments)

***args** collects extra positional arguments.

It gathers all the values that are passed by position (order) — not by name — into one tuple.

**Example**

```python
def add_numbers(*args):
    print(args)

add_numbers(1, 2, 3, 4)
```

args becomes a tuple.

We can pass any number of values.

**Practical Example**

```python
def add_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total

print(add_numbers(10, 20, 30))
```

## 2. ***kwargs** (Variable Keyword Arguments)

***kwargs** collects extra named (keyword) arguments into a dictionary.

**Example**

```python
def print_info(**kwargs):
    print(kwargs)

print_info(name="John Doe", age=44, city="New York")
```

**Practical Example**

```python
def user_info(**kwargs):
    for key, value in kwargs.items():
        print(key, value)

user_info(name="John Doe", age=44, city="New York")
```

# Decorators In Python

Easy Definition: A decorator is just a function that **wraps another function** to add, change, or extend its behavior **without modifying the original function’s code**.

Conceptua Definition: A decorator is a **callable** that receives a **function**, adds extra functionality, and returns a **modified callable**.

**What is a callable ?**

A callable is any object that you can call using parentheses ().

**Basic Example of a Decorator (Without @)**

```python
def my_decorator(func):
    def wrapper():
        print("Before function runs")
        func()
        print("After function runs")
    return wrapper

def say_hello():
    print("Hello!")

# now the functionality of say_hello function will change
say_hello = my_decorator(say_hello)

say_hello()
```

**Basic Example of a Decorator (With @)**

```python
def my_decorator(func):
    def wrapper():
        print("Before")
        func()
        print("After")
    return wrapper

# recommended way to use decorators
@my_decorator
def greet():
    print("Hi!")

greet()
```

**Another Example of a Decorator with Arguments (Very Important Because functions often recieves parameters)**

```python
def logger(func):
    def wrapper(*args, **kwargs):
        # *args, **kwargs allow the decorator to work with any function signature.
        print("Function called")
        return func(*args, **kwargs)
    return wrapper

@logger
def add(a, b):
    return a + b

print(add(2, 3))
```

**A Real World Example of a Decorator which finds out a functions's execution time**

```python
import time

# we can call this function a 'timing decorator'
def time_it(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("Execution time:", end - start)
        return result
    return wrapper

@time_it
def slow_task():
    time.sleep(1)
    print("Task done")

slow_task()
```

# File Handling in Python

File handling means working with files stored on disk — like reading data, writing data, or updating content.

Python provides built-in functions to open, read, write, and manage files.

## 1. Opening a File

**We use the open() function.**

```python
file = open("dummy_text.txt", "r")
# syntax : open(filename, mode)
```

**Common Modes**

## 2. Reading Files

**Read entire file**

```python
f = open("dummy_text.txt", "r")
data = f.read()
print(data)
f.close()
```

**Read one line**

```python
f = open("dummy_text.txt", "r")
line = f.readline()
print(line)
f.close()
```

```python
f = open("dummy_text.txt", "r")
lines = f.readlines()
print(lines)
# 1st line
print(lines[0])
# 2nd line
print(lines[1])
f.close()
```

## 3. Writing Files

**Write (overwrites existing content)**

```python
# Imp: If we provide an existing file, then it will overwrite it otherwise it will create a new file
f = open("example.txt", "w")
f.write("Hello World")
f.close()
```

## 4. Append (adds content at the end)

```python
f = open("example.txt", "a")
f.write("\nNew line added")
f.close()
```

## Important

Earlier, we manually closed files using f.close(). Closing a file after completing operations is important because an open file continues to occupy system resources such as memory and file handles. If a programmer forgets to close the file, it can lead to unnecessary memory usage, resource leaks, or even file corruption in some cases.

To solve this problem, Python provides the 'with' statement, which uses a context manager. When a file is opened using 'with', Python automatically handles cleanup and closes the file safely once the block of code finishes execution — even if an error occurs.

**Example**

```python
with open("example.txt", "r") as f:
    data = f.read()
    print(data)
```
