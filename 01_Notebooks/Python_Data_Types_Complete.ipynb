# Python Data Types – Complete Introduction

This notebook provides a **clear and beginner-friendly explanation of Python data types**. Each data type is explained with **definitions, properties, and basic examples**.

Understanding data types is fundamental for writing correct, efficient, and readable Python code.

## 1. What are Data Types?

A **data type** defines:
- What kind of value a variable can hold
- What operations can be performed on it

Python is **dynamically typed**, meaning you do not need to explicitly declare the data type.

```python
x = 10
x = "Python"  # type can change at runtime
print(x)
```

## 2. Classification of Python Data Types

Python data types can be broadly classified into:

- **Numeric Types** → int, float, complex, bool
- **Sequence Types** → str, list, tuple
- **Set Types** → set, frozenset
- **Mapping Type** → dict
- **None Type** → None

## 3. Numeric Data Types

### 3.1 int (Integer)
Stores whole numbers with unlimited precision.

```python
a = 100
b = -25
print(a, type(a))
print(b, type(b))
```

### 3.2 float (Floating Point)
Stores decimal numbers.

```python
x = 3.14
y = 2.0
print(x, type(x))
print(y, type(y))
```

### 3.3 complex (Complex Numbers)
Used to represent numbers in the form `a + bj`.

```python
z = 3 + 4j
print(z)
print(z.real, z.imag)
```

### 3.4 bool (Boolean)
Represents truth values: `True` or `False`.
It is a subclass of `int`.

```python
is_active = True
print(is_active)
print(True + True)
```

## 4. Sequence Data Types

### 4.1 str (String)
Immutable sequence of characters.

```python
name = "Python"
print(name)
print(name[0])
```

### 4.2 list
Ordered, mutable collection of items.

```python
numbers = [1, 2, 3]
numbers.append(4)
print(numbers)
```

### 4.3 tuple
Ordered, immutable collection of items.

```python
point = (10, 20)
print(point)
```

## 5. Set Data Types

### 5.1 set
Unordered collection of unique elements.

```python
unique_nums = {1, 2, 2, 3}
print(unique_nums)
```

### 5.2 frozenset
Immutable version of a set.

```python
fs = frozenset([1, 2, 3])
print(fs)
```

## 6. Mapping Data Type

### dict (Dictionary)
Stores key-value pairs.

```python
student = {
    "name": "Alex",
    "age": 21
}
print(student)
print(student["name"])
```

## 7. None Type

`None` represents the absence of a value.

```python
result = None
print(result)
print(type(result))
```

## 8. Checking Data Types

```python
x = 10
y = "Python"
print(type(x))
print(type(y))
```

## 9. Mutability vs Immutability

**Mutable:** list, dict, set

**Immutable:** int, float, bool, str, tuple, frozenset
