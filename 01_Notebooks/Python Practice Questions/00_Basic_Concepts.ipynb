
# Python Practice Questions on **Data Types**, **Conditionals**, **Loops**

## Basic Variable & Data Type Code

1. Write a program to store your name and age in variables and print them.

2. Write a program to take two numbers and print their sum, difference, product, and division.

3. Write a program to convert a float value to an integer and print the result.

4. Write a program to swap two numbers using a temporary variable.

5. Write a program to check whether two variables have the same value.

## Conditionals

6. Write a program to check whether a number is positive, negative, or zero.

7. Write a program to check whether a number is even or odd.

8. Write a program to check whether a number lies between 10 and 50 (inclusive).

9. Write a program to find the largest of three numbers using if-elif-else.

10. Predict the output:
```
x = 15
if x < 10:
    print("A")
elif x < 20:
    print("B")
else:
    print("C")
```

---

## Loops

11. Write a program to print numbers from 1 to 10 using a for loop.

12. Write a program to print numbers from 1 to 10 using a while loop.

13. Write a program to print all even numbers between 1 and 30.

14. Write a program to calculate the sum of numbers from 1 to n (assume n is already stored in a variable).

15. Predict the output:
```
for i in range(1, 6):
    if i == 4:
        break
    print(i)
```

## Intermediate Loop Logic

16. Write a program to count the number of digits in a given number.

17. Write a program to reverse a number (example: 456 → 654).

18. Write a program to print the multiplication table of a number stored in a variable.

19. Write a program to print all numbers from 1 to 50 that are divisible by 3 or 5.

20. Predict the output:
```
i = 0
while i < 5:
    i += 1
    if i == 2:
        continue
    print(i)
```
