# Python Practice Questions (Basic to Intermediate)

1. Write a Python program to create a list of 5 numbers and print their sum.

2. Given a list:
   numbers = [10, 20, 30, 40, 50]
   Print the first and last element using indexing.

3. Write a program to count the number of vowels in a string. Example: "hello world"

4. Write a loop that prints numbers from 1 to 20. If number is divisible by 3 → print "Fizz" otherwise print the number.

5. Write a function greet(name) that returns: "Hello <name>"

6. Given nums = [1, 2, 3, 4, 5, 6], create a new list containing only even numbers using a loop and list comprehension.

7. Reverse a string without using built‑in reverse() or [::-1].

8. Write a function that checks whether a string is a palindrome.

9. Given nums = [1, 2, 3, 4, 5], create a new list containing squares of each number using list comprehension.

10. Write a function that takes a list and returns the largest element without using max().

11. Given nums = [1, 2, 2, 3, 4, 4, 5], remove duplicates without using set().

12. Write a program to count frequency of each character in a string. Example: "banana".

13. Using list comprehension, create a list of numbers from 1 to 50 divisible by both 3 and 5.

14. Flatten the 2D list matrix = [[1,2,3],[4,5,6],[7,8,9]] using list comprehension.

15. Write a function that makes vowels uppercase and consonants lowercase in a string.

16. Write a function that rotates a list to the right by k positions.

17. Reverse the order of words in: "python is very powerful".

18. Write a function that finds the second largest element in a list.

19. From words = ["apple","banana","cat","dog","elephant"], return words whose length is greater than 3 using list comprehension.

20. Write a function that squares even numbers and cubes odd numbers using list comprehension.
