# What is Numpy ?

Numpy (Numerical Python) is a core python library used for fast numerical computations, working with arrays and matrices, linear algebra and scientific computations. It provides a powerful object call ndarray (n-dimensional array) which is the heart of numpy.

# Why we need Numpy ?

We need numpy because there are multiple problems with python list data type. Python list can store elements of different data types, and they store the reference of the python object as items therefore python lists are not memory efficient and they are slow in mathematical operations. Python list do not supports vectorised computations.

what are **vectorised computations** ?

Vectorized computation in NumPy is a technique that allows you to perform mathematical operations on entire arrays—or "vectors"—of data at once, instead of looping through each element individually.

Numpy array operations are 10x-100x faster than the python list operations, because it is written in C internally which uses contiguous memory. 

**What contiguous memory means ?**

Contiguous memory means data is stored next to each other in a single continuous block of memory.

Numpy arrays store homogeneous (same) data types only, this allows efficient memory usage and faster computations. The numpy array stores the actual elements (not reference) in continuous memory block. The most important advantage is that numpy arrays supports vectorised computations which is critical in Neural Networks, Linear Regression, Matrix Multiplication. Numpy supports multidimensional arrays ( 1D , 2D, 3D (tensors) , n-dimensional array ) which are the foundation of images(3D array), videos(4D array), and neural networks. 

Almost all Machine Learning libraries are built on top of numpy like scikit-learn, TensorFlow, PyTorch, Pandas, etc.

**Interview Definition**

NumPy is a Python library for efficient numerical computation. It provides the ndarray object, which stores homogeneous data in contiguous memory blocks. This allows fast vectorized operations using optimized C backend, making it much faster and more memory efficient than Python lists. It is widely used in machine learning and scientific computing.

**What is ndarray in numpy ?**

ndarray (n-dimensional array) is the core object in Numpy. It is a grid of values stored in contiguous memory. All elements have same data types (Homogeneous). ndarray can be 1D, 2D, 3D or n-dimensional.

# Comparing time consumption between python list and numpy array

## Python List

```python
import time

a = [i for i in range(10000000)]
b = [i for i in range(10000000,20000000)]

c = []

start_time = time.time()

for i in range(len(a)):
    c.append(a[i] + b[i])
    
print(time.time() - start_time)
```

## Numpy Array

```python
import time 

a = np.arange(10000000)
b = np.arange(10000000,20000000)

start_time = time.time()
c = a + b
print(time.time() - start_time)
```

We can clearly observe that the numpy array operation is approximately 30-40 times faster than the python list operation.

# Broadcasting

Broadcasting allows NumPy to Perform operations on arrays of different shapes automatically
without manually reshaping them.

It “stretches” the smaller array to match the bigger one —
without actually copying data.

```python
# Simple Example
a = np.array([10,20,30,40])
print(a + 10) # 10 behaves like [10,10,10,10]
```

Here:

10 behaves like [10,10,10,10]

But NumPy does NOT create that array

It broadcasts it internally

```python
# 2D Example
a = np.array([[1,4,2],[9,5,9]])
b = np.array([5,7,2])

a+b
```

Here, b is treated like : 
[[5,7,2]
 [5,7,2]]

Numpy broadcasted it row-wise

## Broadcasting Rules

Numpy compares shapes from right to left.

Two arrays are compatible if one of the following condition is met:
1. The dimensions of both the arrays is same.
2. Either row or column is 1.

Examples: 
1. (2,3) (3,)   --> missing number is treated as 1  --> (2,3) (1,3) --> Now 1 can stretch upto 2 --> (2,3) (2,3) --> Same
2. (2,3) (2,1)  --> 1 can strect upto 3 --> (2,3) (2,3) --> Same

**Example** (Column Broadcasting)

```python
# dimension : (2,3)
a = np.array([[1,2,3],
              [4,5,6]])

# dimension : (2,1)
b = np.array([[10],
              [20]])

a + b
```

**Example** (Invalid)

```python
# (2,3)
a = np.array([[1,2,3],
              [4,5,6]])
# (2,2)
b = np.array([[1,2],
              [4,5]])

# a + b
```

**Example** (Scaler Broadcasting)

```python
a = np.array([[1,2,3],
              [4,5,6]])

a * 5 # Scaler behaves like full size array
```

# Mathematical Formulas with Numpy

## 1. Sigmoid

Sigmoid function is widely used in logistic regression and binary classification.

Formula : sigmoid = 1 / (1 + np.exp(-x))

```python
def sigmoid(arr):
    ans = 1 / (1 + np.exp(-arr))
    return ans

a = np.arange(50000)

result = sigmoid(a)

print(result)
```

## 2. Mean Squared Error (MSE)

Mean Squared Error is a regression loss function that computes the average of squared differences between actual and predicted values. It penalizes larger errors more heavily due to squaring.

```python
y = np.array([3, -0.5, 2, 7])
y_pred = np.array([2.5, 0.0, 2, 8])

mse = np.mean((y - y_pred) ** 2)
print(mse)
```

MSE is used in: 

1. Linear Regression

2. Neural Networks (regression tasks)

3. Model evaluation

# Working with Missing Values

In NumPy, missing values are usually represented using:

np.nan (Not a Number)

Sometimes masked arrays

np.nan means:

A floating-point value representing missing or undefined data.

```python
a = np.array([1, 2, np.nan, 4])
print(a)
```

```python
# Important point
np.nan == np.nan
# Because NaN is not equal to anything — even itself.
```

```python
# Detect Missing Values
np.isnan(a)
```

```python
# Get missing values
a[np.isnan(a)]
```

```python
# Remove missing values
a[~np.isnan(a)]
```

```python
# Replace missing values
a[np.isnan(a)] = 0
```
