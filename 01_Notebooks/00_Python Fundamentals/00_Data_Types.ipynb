# Python Data Types – Complete Introduction

This notebook provides a **clear and beginner-friendly explanation of Python data types**. Each data type is explained with **definitions, properties, and basic examples**.

Understanding data types is fundamental for writing correct, efficient, and readable Python code.

## 1. What are Data Types?

A **data type** defines:
- What kind of value a variable can hold
- What operations can be performed on it

Python is **dynamically typed**, meaning you do not need to explicitly declare the data type.

```python
a = 10.04
print(type(a))
print(id(a))
```

```python
a = int(input())
type(a)
```

```python
x = 10
x = "Python"    # type can change at runtime
a = 2.9
b = 5
print(x)
print(type(x))
print(id(a)) #9760

```

## 2. Classification of Python Data Types

Python data types can be broadly classified into:

- **Numeric Types** → int, float, complex, bool
- **Sequence Types** → str, list, tuple
- **Set Types** → set, frozenset
- **Mapping Type** → dict
- **None Type** → None

## 3. Numeric Data Types

### 3.1 int (Integer)
Stores whole numbers with unlimited precision.

```python
a = 100
b = -25
c = None
print(a, type(a))
print(b, type(b))
print(type(c))
```

### 3.2 float (Floating Point)
Stores decimal numbers.

```python
x = 3.14
y = 2.0
print(x, type(x))
print(y, type(y))
```

### 3.3 complex (Complex Numbers)
Used to represent numbers in the form `a + bj`.

```python
z = 3 + 4j
print(z)
print(z.real, z.imag)
```

### 3.4 bool (Boolean)
Represents truth values: `True` or `False`.
It is a subclass of `int`.

```python
is_active = True
print(is_active)
print(True + True)
print(True + False) 
```

## 4. Sequence Data Types

### 4.1 str (String)
Immutable sequence of characters.

```python
name = "Python"
print(type(name))
print(name)
print(name[4])
print(len(name))
name.upper().count('P')
```

### 4.2 list
Ordered, mutable collection of items.

```python
# List is like array
print([1,3,4,5,'Salman'])

lst = [1,2,3,4,'Salman','Seraj',2.45]

lst.reverse() 
a = lst.append('Danish') # Doesnt return new list
lst
```

### 4.3 tuple
Ordered, immutable collection of items.

```python
#Tuple
print((1,2,3,4,'Hell0')) # brackets are different. There are other difference from list

point = (10, 20, 30)
print(type(point))
```

## 5. Set Data Types

### 5.1 set
Unordered collection of unique elements.

```python
unique_nums = {1, 2, 2, 3}
print(unique_nums)
```

## 6. Mapping Data Type

### dict (Dictionary)
Stores key-value pairs.

```python
#dict

print({"Name":"salman", "Age":40}) # like key:value pair

student = {
    "name": "Alex",
    "age": 21,
    "Location": "Hyderabad"
}
#print(student)
print(student["name"])
```

```python
#What does it mean?
'name' in student.keys()
```

```python
people = [
    {"name": "Alice", "age": 25, "city": "New York"},
    {"name": "Bob", "age": 30, "city": "London"},
    {"name": "Charlie", "age": 22, "city": "Paris"}
]

#people[2].keys()
#people[0]["name"]
```

```python
for person in people:
    print(person.get('name'))
    if 'age' in person.keys():
        print("Age is present")
```

## 7. None Type

`None` represents the absence of a value.

```python
result = None
print(result)
print(type(result))
```
