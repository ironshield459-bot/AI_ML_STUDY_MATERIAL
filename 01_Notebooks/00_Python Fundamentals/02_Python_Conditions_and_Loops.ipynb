# Python Conditions & Loops – Important Concepts Only

This notebook covers **only the most important and commonly used concepts** of **conditions and loops** in Python.

These concepts are essential for:
- Control flow
- Backend logic
- Data processing
- Interviews

## 1. Conditional Statements (if / elif / else)

Conditions allow your program to make decisions based on boolean expressions.

```python
x = 10

if x > 0:
    print("Positive")
elif x == 0:
    print("Zero")
else:
    print("Negative")
```

### Comparison Operators
`==  !=  >  <  >=  <=`

```python
a, b = 5, 10
print(a < b)
print(a == b)
```

### Logical Operators
`and`, `or`, `not`

```python
age = 21
is_student = True

print(age > 18 and is_student)
print(not is_student)
```

## 2. Truthy & Falsy Values (Very Important)

Falsy values include:
- `0`, `0.0`
- `''`
- `[]`, `{}`, `set()`
- `None`

```python
values = [0, "", [], None, "Python"]

for v in values:
    if v:
        print(v, "is truthy")
    else:
        print(v, "is falsy")
```

## 3. Conditional Expression (Ternary Operator)

Short form of if-else.

```python
x = 5
result = "Even" if x % 2 == 0 else "Odd"
print(result)
```

## 4. for Loop

Used to iterate over sequences (list, tuple, string, dict).

```python
for i in range(3):
    print(i)
```

### Looping Over Collections

```python
data = [10, 20, 30]
for item in data:
    print(item)
```

## 5. while Loop

Used when the number of iterations is not known beforehand.

```python
count = 0
while count < 3:
    print(count)
    count += 1
```

## 6. break, continue, pass

- `break` → exits loop
- `continue` → skips current iteration
- `pass` → placeholder

```python
for i in range(5):
    if i == 2:
        continue
    if i == 4:
        break
    print(i)
```

```python
if True:
    pass
```

## 7. else with Loops

`else` runs only if loop completes normally (no break).

```python
for i in range(3):
    print(i)
else:
    print("Loop completed")
```

```python
for i in range(3):
    if i == 1:
        break
else:
    print("Will not run")
```

## 8. Nested Loops (Basic Use)

```python
for i in range(2):
    for j in range(2):
        print(i, j)
```

## 9. enumerate() and zip() (Very Common)

Used to loop with index or multiple sequences.

```python
items = ['a', 'b', 'c']
for idx, val in enumerate(items):
    print(idx, val)

nums = [1, 2, 3]
letters = ['x', 'y', 'z']
for n, l in zip(nums, letters):
    print(n, l)
```
