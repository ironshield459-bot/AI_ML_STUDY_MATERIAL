# Python Tuples – Important & Commonly Used Concepts

## 1. What is a Tuple?

A **tuple** is:
- Ordered
- Immutable
- Indexed
- Heterogeneous

**Key properties:**
- Allows duplicate values
- Faster than lists for iteration
- Hashable (if elements are hashable)
- Commonly used for fixed collections

```python
t = (10, "python", 3.14, True)
print(t)
```

## 2. Creating Tuples

Different common ways to create tuples.

```python
t1 = ()              # empty tuple
t2 = (10,)           # single-element tuple (comma is required)
t3 = tuple([1, 2, 3])
t4 = tuple("abc")

print(t1)
print(t2)
print(t3)
print(t4)
```

⚠️ **Important:** `(10)` is NOT a tuple, it is an `int`.

```python
print(type((10)))
print(type((10,)))
```

## 3. Indexing & Slicing

Tuples support indexing and slicing just like lists.

```python
t = (10, 20, 30, 40, 50)

print(t[0])
print(t[-1])
print(t[1:4])
print(t[::-1])
```

## 4. Immutability (Very Important Concept)

- Tuple elements **cannot be reassigned**
- This makes tuples safer for read-only data

```python
t = (10, 20, 30)
# t[0] = 100  # ❌ TypeError
```

### Mutability Inside a Tuple (Common Confusion)

If a tuple contains **mutable objects**, those objects can still be modified.

```python
t = ([1, 2], [3, 4])

t[0].append(99)
print(t)
```

## 5. Tuple Methods (Only Two)

Tuples intentionally have very few methods.

```python
t = (1, 2, 2, 3)

print(t.count(2))
print(t.index(2))
```

## 6. Built-in Functions Used with Tuples

```python
t = (10, 20, 30)

print(len(t))
print(max(t))
print(min(t))
print(sum(t))
print(sorted(t))
```

## 7. Tuple Packing & Unpacking (Very Important)

Used heavily in function returns and variable assignment.

```python
# Packing
t = 1, 2, 3
print(t)

# Unpacking
a, b, c = t
print(a, b, c)

# Extended unpacking
x, *y, z = (1, 2, 3, 4, 5)
print(x, y, z)
```

## 8. Tuple vs List (Why Tuples Exist)

```python
lst = [1, 2]
tpl = (1, 2)

# lst[0] = 100  # allowed
# tpl[0] = 100  # ❌ not allowed

print(lst)
print(tpl)
```

## 9. Hashability & Dictionary Keys

Tuples can be used as dictionary keys if they contain only immutable elements.

```python
points = {(1, 2): "A", (3, 4): "B"}
print(points[(1, 2)])
```
