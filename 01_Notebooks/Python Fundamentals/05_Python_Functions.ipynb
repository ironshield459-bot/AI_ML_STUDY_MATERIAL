# Python Functions – Important Concepts Only

This notebook covers **only the most important and commonly used concepts of Python functions**.

Functions are fundamental for:
- Writing reusable code
- Organizing logic
- Backend development
- Interviews

## 1. What is a Function?

A **function** is a reusable block of code that:
- Performs a specific task
- Runs only when called
- Can accept inputs (parameters)
- Can return output

```python
def greet():
    print("Hello, Python")

# function call
greet()
```

## 2. Parameters & Arguments (Very Important)

- **Parameter** → variable in function definition
- **Argument** → value passed during function call

```python
def greet(name):
    print(f"Hello, {name}")

greet("Alex")
```

## 3. return Statement

- Sends value back to caller
- Stops function execution
- Default return value is `None`

```python
def add(a, b):
    return a + b

result = add(3, 4)
print(result)
```

```python
def demo():
    pass

print(demo())
```

## 4. Default Arguments

Used to provide fallback values.

```python
def power(base, exp=2):
    return base ** exp

print(power(3))
print(power(3, 3))
```

## 5. Keyword Arguments

Arguments passed using parameter names.

```python
def student(name, age):
    print(name, age)

student(age=21, name="Danish")
```

## 6. *args (Variable Positional Arguments)

Allows passing multiple positional arguments.

```python
def total(*nums):
    return sum(nums)

print(total(1, 2, 3, 4))
```

## 7. **kwargs (Variable Keyword Arguments)

Used to pass named arguments as a dictionary.

```python
def user_info(**data):
    for k, v in data.items():
        print(k, v)

user_info(name="Alex", age=21)
```

## 8. Order of Arguments 

`positional → default → *args → **kwargs`

```python
def demo(a, b=2, *args, **kwargs):
    print(a, b, args, kwargs)

demo(1, 3, 4, 5, x=10)
```

## 9. Function Scope (LEGB Rule – Important)

- Local
- Enclosing
- Global
- Built-in

```python
x = 10

def show():
    x = 5
    print(x)

show()
print(x)
```

## 10. Lambda Functions (Short Functions)

Used for small, one-line functions.

```python
square = lambda x: x * x
print(square(5))
```

## 11. Docstrings

Used to document functions.

```python
def add(a, b):
    """Returns sum of two numbers"""
    return a + b

print(add.__doc__)
```
