# Python Fundamentals: Numbers & Strings (Detailed Guide)

This notebook provides a **detailed yet practical introduction** to Python **Numbers** and **Strings**. The content is curated from standard Python reference notes and focuses on concepts **commonly used in real-world Python, backend development, and interviews**.

## 1. Python Number Types

Python has several built-in numeric types:

- **int** → Whole numbers with unlimited precision
- **float** → Decimal numbers (IEEE-754 double precision)
- **bool** → Boolean values (`True`, `False`), subclass of `int`

**Key property:** All numeric types in Python are **immutable**.

```python
a = 10
b = 3.14
c = True

print(a, type(a))
print(b, type(b))
print(c, type(c))
print(True + True)  # bool behaves like int
```

## 2. Integers (`int`)

- Can store arbitrarily large numbers
- Can be created from numbers or strings
- Supports base conversion (binary, octal, hexadecimal)

### Arithmetic Operations

`+  -  *  /  //  %  **` are heavily used in real code.

```python
a, b = 5, 2

print(a / b)    # true division
print(a // b)   # floor division
print(a % b)    # modulus
print(a ** b)   # power
print(a + b)
```

### Built-in Numeric Helpers

```python
print(abs(-10))
print(pow(2, 3))
print(pow(2, 3, 5))  # modular power

```

## 3. Floating Point Numbers (`float`)

- Used for decimal values
- Stored internally in binary form
- Can introduce precision issues

```python
x = 3.14
y = float("2.5")

print(x, y)
print(x.is_integer())
print(5.0.is_integer())
```

### Rounding & Precision Issues

Python uses **banker's rounding** and binary floating-point representation.

```python
print(round(3.14159, 2))
print(round(2.5))
print(0.1 + 0.2)
```

## 4. Math Module (Common Functions)

```python
import math

print(math.sqrt(16))
print(math.floor(3.7))
print(math.ceil(3.2))
print(math.pi)
```

## 5. Type Conversion & Truthy/Falsy Values

- `int()` truncates, it does NOT round
- `bool()` follows truthy/falsy rules

```python
print(int(3.9))
print(float(10))
print(bool(0))
print(bool(""))
print(bool("python"))
print(bool(""))


value ='some value'
if value:
    print("There is a value")
```

## 6. Python Strings – Fundamentals

Strings are **immutable sequences of characters**.
They are one of the most used data types in Python.

```python
s1 = "Python"
s2 = 'Backend'
print(s1,s2)
```

## 7. Indexing & Slicing (Very Important)

- Index starts at 0
- Negative index starts from the end

```python
text = "PythonBackend"

print(text[0])
print(text[-1])
print(text[0:6])
print(text[6:])
```

## 8. Common String Methods (Daily Use)

```python
msg = "python backend is an important skill in todays time "

print(id(msg))

newstring = msg.upper()
print(newstring)
print(id(newstring))
print(msg.lower())
print(msg.strip())
print(msg.replace("backend", "developer"))
print(msg.split(" "))
```

## 9. Searching & Counting in Strings

```python
word = "banana"
print(word.find("na"))
print(word.rfind("na"))
print(word.count("a"))
```

## 10. Prefix & Suffix Checks

```python
filename = "data.csv"
print(filename.startswith("data"))
print(filename.endswith(".csv"))
```

## 11. Splitting & Joining Strings (Very Important)

```python
data = "2026-01-30"
parts = data.split("-")
print(parts)
print("/".join(parts))
```

## 12. String Formatting

**f-strings** are the preferred modern approach.

```python
name = "Alex"
score = 95
print(f"{name} scored {score}")
```

## 13. String Validation Methods

```python
print("123".isdigit())
print("abc".isalpha())
print("1_var".isidentifier())
```
