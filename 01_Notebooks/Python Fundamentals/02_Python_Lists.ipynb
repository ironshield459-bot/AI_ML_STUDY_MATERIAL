# Python Lists – Important & Commonly Used Concepts

This notebook focuses on **Python lists**, covering **only the most important and commonly used concepts**.

Lists are one of the most frequently used data types in Python and are critical for:
- Backend development
- Data processing
- Interviews
- Real-world problem solving

## 1. What is a List?

A **list** is an:
- Ordered collection
- Mutable (can be changed)
- Indexed
- Heterogeneous (can store different data types)

**Important properties:**
- Allows duplicates
- Stores references to objects
- Implemented internally as a dynamic array

```python
lst = [10, "python", 3.14, True]
print(lst)
```

## 2. Creating Lists

Lists can be created in multiple common ways.

```python
a = []
b = list()
c = list("abc")
d = list(range(5))

print(a)
print(b)
print(c)
print(d)
```

```python
a = "Learning Python"
list(a)
```

## 3. Indexing & Slicing (Very Important)

- Index starts from 0
- Negative index starts from the end
- Slicing returns a **new list**

```python
nums = [10, 20, 30, 40, 50]

print(nums[0])
print(nums[-1])
print(nums[:])
print(nums[0:5:2])
print(nums[::-1])
```

## 4. Commonly Used List Methods

These methods are used **daily in real Python code**.

```python
# append
lst = [1, 2]
lst.append(3)
print(lst)

# extend
lst.extend([4, 5])
print(lst)

# insert
lst.insert(1, 100)
print(lst)
```

### append vs extend (Very Important Difference)

```python
a = [1, 2]
a.append([3, 4])
print(a)

b = [1, 2]
b.extend([3, 4])
print(b)
```

## 5. Removing Elements from Lists

```python
lst = [1, 2, 2, 3]

lst.remove(2)  # removes first occurrence
print(lst)

last = lst.pop()
print("popped element", last)
print(lst)
lst.clear()
print(lst)

lst.sort()
lst
```

```python
lst = [3,4,6,2,9,4,1,0,4]
lst.sort()
lst
```

## 6. Searching & Counting Elements

```python
lst = [10, 20, 20, 30]
print(lst.index(20))
print(lst.count(20)) #no. of occurence
```

## 7. Sorting & Reversing Lists

```python
nums = [3, 1, 4, 2]
nums.sort()
print(nums)

words = ["apple", "Banana"]
words.sort(key=str.lower)
print(words)

nums.reverse()
print(nums)
```

## 8. Copying Lists (Shallow Copy – Important)

```python
a = [1, 2]
b = a.copy()

b.append(3)
print(a)
print(b)
```

### Shallow Copy Pitfall

```python
a = [[1, 2], [3, 4]]
b = a.copy()

b[0].append(99)
print(a)
print(b)
```

## 9. Built-in Functions with Lists

```python
nums = [1, 2, 3, 4]
print(len(nums))
print(max(nums))
print(min(nums))
print(sum(nums))
print(sorted(nums, reverse=True))
```

## 10. Membership & Iteration

```python
nums = [1, 2, 3]

print(2 in nums)

for n in nums:
    print(n)
```

```python
[i*i for i in nums if i > 2]
```

## 11. List Comprehensions (Core Python Skill)

Compact way to create lists.

```python
squares = [x * x for x in range(5)]
print(squares)

# with condition
evens = [x for x in range(10) if x % 2 == 0]
print(evens)
```

```python
#Write a program to multiply two numbers a and b without using *

a, b = 5, 6

sum([a for i in range(1,b+1)])
    
```
