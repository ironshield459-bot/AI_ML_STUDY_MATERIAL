```python
from langchain_community.document_loaders import PyMuPDFLoader
#from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import Chroma

# 1. Load PDF
loader = PyMuPDFLoader("C:/Users/ruhin/MyProjects/AI_ML_STUDY_MATERIAL/01_Notebooks/01_SampleData/2604.27096v1.pdf")  # use forward slashes
documents = loader.load()


# 3. Create embeddings (Ollama)
embeddings = OllamaEmbeddings(
    model="nomic-embed-text"  # or mxbai-embed-large
)

# 4. Store in ChromaDB
vectorstore = Chroma.from_documents(
    documents=docs,
    embedding=embeddings,
    persist_directory="./chroma_db"  # saves locally
)

# 5. Persist to disk
vectorstore.persist()

print("✅ PDF embedded and stored in ChromaDB")
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

results = retriever.invoke("What is this PDF about?")

for doc in results:
    print(doc.page_content[:200])
```
